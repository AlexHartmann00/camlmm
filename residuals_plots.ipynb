```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = pd.read_csv('project_data/residuals_lmm_vader_6hr.csv')
data
```

```python
bins = np.linspace(-0.1, 0.1,25)

fig,ax = plt.subplots(figsize=(16,6))

ax.hist(data.loc[data.anxiety == 1]["residuals"], alpha  =0.5, bins = bins, density=True, label='Anxiety')
ax.hist(data.loc[data.anxiety == 0]["residuals"], alpha  =0.5, bins = bins, density=True, label='Depression')
ax.legend()
ax.set_title('Residual distribution of LMM for 6-hour data')
ax.set_ylabel('Density')
ax.set_xlabel('Residuals')
plt.savefig('figures/lmm_6hr_vader_residuals.png', dpi=400, bbox_inches='tight')
```
